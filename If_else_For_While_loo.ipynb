```python
# 4. Factorial Calculator:Write a program to calculate the factorial of a number.
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial is:", fact)
```

```python
# 1.Countdown Timer:Write a program to count down from a given number to 0 using a while loop.
num = int(input())

while num != 0:
    num = num - 1
    print (num)
```

```python
#2. Guess the Number:Write a guessing game where the user keeps guessing a predefined number until they get it right.
secret = 55
guess= 0
while guess != secret:
    guess = int (input())

    if guess > secret :
        print('value greater than number')
    elif guess< secret:
        print('Value is smaller than the number')
    else:
        guess == secret
        print('Correct Guess')
    
```

```python
# 4. Reverse Digits: Input a number and reverse its digits using a while loop.
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10          # get last digit
    reverse = reverse * 10 + digit
    num = num // 10           # remove last digit

print("Reversed number:", reverse)
```

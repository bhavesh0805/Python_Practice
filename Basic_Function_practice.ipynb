```python
# Vowel Counter: Create a function count_vowels(word) that loops through a string and prints the number of vowels found.
def count_vowels(word):
    count = 0
    for char in word.lower():   # Convert to lowercase once
        if char in "aeiou":     # Only need to check lowercase vowels
            count += 1
    print(f"'{word}' contains {count} vowel(s)")
    return count
```

```python
count_vowels('bhavesh')
```

```python
# Default Arguments: Create a function power(base, exp=2) so that if no exponent is provided, it automatically squares the number.
exp1= int(input())
def power(base, exp=2):
    if exp1 != 2:
        return base**exp1
    else:
        return base**2
```

```python
power(3,2)
```

```python
# Simple Calculator: Create a function calc(a, b, operation) where operation can be "add", "sub", "mul", or "div".
def calc(a, b, operation):
    if operation == 'add':
        return  a+b
    elif operation == 'sub':
        return  a-b
    elif operation == 'mul':
        return a*b
    elif operation == 'div':
        return a/b
```

```python
calc(5,6,'div')
```

```python
# Nested Logic: Write a function that takes a list of ages and uses the even_odd logic from your notebook to count how many even-numbered ages exist.
l=[]
def ages(l):
    even = 0
    odd = 0
    for i in l:
        if i % 2 == 0:
            even = even + 1
            
        elif i % 2 != 0:
            odd = odd + 1 
    print(f'odd ages are {odd}')
    print(f'Even ages are {even}')
```

```python
ages([24,58,55,37,97,25,46])
```
